Flow control in Python allows our programs to make decisions based on certain conditions. This is achieved using conditional structures that evaluate boolean expressions and, depending on the result, execute specific blocks of code. Below we will explore the main conditional structures in Python.

---

## **Conditionals**

Conditionals are fundamental for controlling program flow. They allow different code fragments to run depending on whether a condition is true or false. To understand them better, let's remember the importance of the boolean data type (`True`, `False`) as well as logical operators (`and`, `or`, `not`) and relational operators (`<`, `<=`, `>`, `>=`, `==`, `!=`).

### **1. `if`**
The `if` block evaluates a condition. If the condition is true, the indented code following the `if` is executed. This is the simplest form of a conditional statement.

### **2. `if ... else`**
The `if ... else` block is used when we need to execute one block of code if the condition is true and another block if the condition is false.

### **3. `if ... elif ... elif ... else`**
When multiple conditions need to be evaluated, the `if ... elif ... elif ... else` block is used. Each `elif` represents an alternative condition. If none of the previous conditions are met, the `else` block is executed.

Therefore, using conditionals in Python allows us to write dynamic and adaptable programs. Depending on the conditions evaluated, we can decide which code blocks to execute, providing greater flexibility in the development of our projects.



```python
#### Example:
print("Example")
x = 10

if x > 5:
    print("x is greater than 5")


# Example 1 using if ... else
print("Example 1")
x = 3

if x > 5:
    print("x is greater than 5")
else:
    print("x is less than or equal to 5")


# Example 2 using if ... elif ... else
print("Example 2")

x = 8

if x > 10:
    print("x is greater than 10")
elif x > 5:
    print("x is greater than 5 but less than or equal to 10")
elif x > 0:
    print("x is positive but less than or equal to 5")
else:
    print("x is negative or zero")

# Example 3 using if ... elif ... elif ... else
print("Example 3")

x = -3

if x > 10:
    print("x is greater than 10")
elif x > 5:
    print("x is greater than 5 but less than or equal to 10")
elif x > 0:
    print("x is positive but less than or equal to 5")
else:
    print("x is negative or zero")
```

```python
### Example 1: Basic `while` usage
# Count from 1 to 5
'''
The while loop starts by checking whether the condition counter <= 5 is true. If it is true,
the code block inside the while loop runs. Then the value of counter is incremented by 1.
The while loop checks again whether counter <= 5 is true and repeats the process.
This process continues until counter <= 5 is false. At that point, the while loop stops and
the program continues with the next line of code after the while loop.
'''
print("Example 1")

counter = 1

while counter <= 5:
    print(f"Iteration {counter}")
    counter += 1

### Example 2: Infinite loop
print("Example 2")

'''
An infinite loop is a loop that has no exit condition.
This means the loop will continue running forever.
# Infinite loop
The condition False is always false, so the while loop will not execute.
'''
print("This loop never ends.")

### Example 3: Using `break` to exit a loop
#The break statement is used to exit a while loop before the loop condition becomes false.
'''
print("Example with break:")
counter = 0
while True:
    print("Counter:", counter)
    if counter == 5:
        print("Reached number 5, loop interrupted!")
        break
    counter += 1
'''
    
### Example 4: Using `continue` to skip to the next iteration
print("Example 4")

counter = 0
while counter < 10:
    counter += 1
    if counter % 2 == 0:
        continue
    print(f"Odd number: {counter}")

### Example 5: Using `else` with a `while` loop
#The else clause on a while loop executes when the loop condition becomes false.
print("Example 5")

counter = 1
while counter <= 3:
    print(f"Iteration {counter}")
    counter += 1
else:
    print("The loop has finished.")

### Example 6: Using `while` with lists
#In this example, a while loop is used to iterate through a list of elements.
print("Example 6")

numbers = [10, 20, 30, 40, 50]
index = 0
while index < len(numbers):
    print(numbers[index])
    index += 1

### Example 7: Using `while` with dictionaries
#In this example, a while loop is used to iterate through a dictionary.
print("Example 7")

grades = {
    "Juan": 90,
    "Maria": 85,
    "Pedro": 88,
    "Ana": 92
}
index = 0
while index < len(grades):
    student = list(grades.keys())[index]
    print(student, grades[student])
    index += 1

### Example 8: Using `while` with strings
print("Example 8")

#In this example, a while loop is used to iterate through a string.
message = "Hello, world!"
index = 0
while index < len(message):
    print(message[index])
    index += 1
```

## **`for ... in`**

The `for` loop in Python is one of the most used control structures because it allows iterating over sequences like lists, tuples, strings, dictionaries, or any iterable object. The `for` loop in Python does not require specifying an index manually, which makes it more concise and easier to use than in many other programming languages.

Loops are powerful tools for automating tasks in our programs. The `while` loop is used when we do not know how many iterations we will need, while the `for` loop is ideal for traversing collections or defined ranges of numbers. In addition, the statements *break* and *continue* offer extra control over loop execution.

```python
# Example for iterating over a list
'''
The range() function is used to generate a sequence of integers.
It can take one, two, or three arguments: range([start], stop[, step]).
'''
# Print the numbers from 0 to 4.
print("Example 1")

for i in range(5):
    print(i)

# Print the numbers from 1 to 5.
print("Example 2")

for i in range(1, 6):
    print(i)

# Print the numbers from 0 to 8 with steps of 2.
print("Example 3")

for i in range(0, 10, 2):
    print(i)

# Print the numbers from 10 to 1 with decrements of 1.
print("Example 4")

for i in range(10, 0, -1):
    print(i)

# Iterate over a list of items.
print("Example 5")

names = ["Alice", "Bob", "Charlie"]

for name in names:
    print(name)

# Iterate over a list of items with indices.
print("Example 6")

names = ["Alice", "Bob", "Charlie"]

for i, name in enumerate(names):
    print(f"Index: {i}, Name: {name}")

# Iterate over a string.
print("Example 7")

text = "Python"

for letter in text:
    print(letter)

# Iterate over a string with indices.
print("Example 8")

text = "Python"

for i, letter in enumerate(text):
    print(f"Index: {i}, Letter: {letter}")

# Iterate over a dictionary.
print("Example 9")

ages = {"Alice": 25, "Bob": 30, "Charlie": 35}

for name, age in ages.items():
    print(f"{name} is {age} years old")
 
```

```python
# Using break and continue in a for loop


# The break statement "ends" the loop when it encounters the Spam item condition
# The continue statement "continues" the loop but skips the item that meets the Spam condition

# Create a list
items_list = ["Bread","Fruit","Spam","Eggs","Cheese"]

print("Condition with break")
for item in items_list:
    if item == "Spam":
        break
    print("Buying:", item)
print("*"*6)

print("Condition with continue")
for item in items_list:
    if item == "Spam":
        continue
    print("Buying:", item)
print("*"*6)
```

### Augmented Assigments

| **Augmented assignment** | **Sample expression** | **Explanation** | **Assigns** |
|--------------------------|-----------------------|-----------------|-------------|
| **Assume:** `c = 3`, `d = 5`, `e = 4`, `f = 2`, `g = 9`, `h = 12` |  |  |  |
| `+=` | `c += 7` | `c = c + 7` | `10` to `c` |
| `-=` | `d -= 4` | `d = d - 4` | `1` to `d` |
| `*=` | `e *= 5` | `e = e * 5` | `20` to `e` |
| `**=` | `f **= 3` | `f = f ** 3` | `8` to `f` |
| `/=` | `g /= 2` | `g = g / 2` | `4.5` to `g` |
| `//=` | `g //= 2` | `g = g // 2` | `4` to `g` |
| `%=` | `h %= 9` | `h = h % 9` | `3` to `h` |


```python
'''
Giving the list of grades, give the average of them. Use for iteration
'''

grades = [98,76,71,87,83,58,78,92,92]
total = 0   # Total of elements
count = 0   # Grade counter


for grade in grades:
    total += 1 # This passes through every item of the list, so it counts the total number of elements
    count += grade

print(count)  
print(total)

# Average operation
average = count / total
print("The average of the list is:", average)
```

## **`match ... case`**
- Multiple selection statement
- Compares subject expression's value to each case's value
- First matching case's suite executes then match..case terminates

```python
grade = "C"

'''
Here match is a variable, but it could also be a function or any other expression. 
The match statement compares the value of the variable with the values specified in each case. 
If a match is found, the corresponding block of code is executed. If no match is found, 
the program continues to the next case until it finds a match or reaches the end of the
match statement.
'''

match grade:
    case "A":
        print("grade is A")
    case "B":
        print("grade is B")
    case "C":
        print("grade is C")
    case "D":
        print("grade is D")
    case "_":  # This case acts as a default case, similar to the else statement in an if-else structure.
        print("grade is F")
```
