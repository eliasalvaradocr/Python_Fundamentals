```python
# Ejemplo 1 Multiply
def multiply(x,y):
    """
    Funcion que realiza una multiplicacion
    """
    result = x * y
    return result

final = multiply(2,3)
print(final)
help(multiply) # En el help muestra los comentarios dentro de la funcion
# Ejemplo 2 Palindrome
def palindrome(string):
    string1 = string.lower() # Coloca en minusculas
    string1 = string1.replace(" ","")
    reverse = string1[::-1]
    return reverse == string1
# Modificar palabra 
palabra = "Race car"
if palindrome(palabra):
    print(palabra, "es palindrome")
else:
    print(palabra,"no es palindrome")

# Ejemplo 3 Fibonacci Numbers
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1 or n ==2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
# Imprime lista de primeros 36 numeros de fibonacci
for i in range(36):
    print(i, fibonacci(i))
```
